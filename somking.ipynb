```python
x = input("Do you smoke? (yes/no): ").strip().lower()

if x == "yes":
    years = int(input("For how many years have you been smoking: "))
    cigs_per_day = int(input("On average, how many cigarettes do you smoke per day: "))
    age = int(input("How old are you: "))

    # Assumption: 1 cigarette ≈ 11 minutes of life lost
    minutes_lost = cigs_per_day * 365 * years * 20
    
    years_lost = minutes_lost / (60 * 24 * 365)

    life_expectancy = 75  # more realistic average
    remaining_years = life_expectancy - age - years_lost

    print("\n--- Health Estimate ---")
    print(f"Estimated years of life lost due to smoking: {years_lost:.2f} years")

    if remaining_years > 0:
        print(f"Estimated remaining life expectancy: {remaining_years:.2f} years")
    else:
        print("Warning: Smoking may have already reduced expected lifespan significantly.")

elif x == "no":
    print("Good choice 👍 Keep it up and maintain a healthy lifestyle!")

else:
    print("ERROR! Please enter only 'yes' or 'no'.")
```

```python
import streamlit as st

st.set_page_config(page_title="Smoking Impact Calculator", page_icon="🚭")

st.title("🚬 Smoking Impact Calculator")
st.write("This app gives a **rough estimate** of how smoking may affect life expectancy.")

smoke = st.radio(
    "Do you smoke?",
    ("Yes", "No")
)

if smoke == "Yes":
    col1, col2 = st.columns(2)

    with col1:
        years = st.number_input(
            "For how many years have you been smoking?",
            min_value=0,
            step=1
        )

        cigs_per_day = st.number_input(
            "Average cigarettes per day",
            min_value=0,
            step=1
        )

    with col2:
        age = st.number_input(
            "Your age",
            min_value=0,
            step=1
        )

    if st.button("Calculate Impact"):
        # Assumption: ~11 minutes of life lost per cigarette
        minutes_lost = cigs_per_day * 365 * years * 11
        years_lost = minutes_lost / (60 * 24 * 365)

        life_expectancy = 75
        remaining_years = life_expectancy - age - years_lost

        st.subheader("📊 Health Estimate")
        st.write(f"**Estimated years of life lost:** `{years_lost:.2f}` years")

        if remaining_years > 0:
            st.success(
                f"**Estimated remaining life expectancy:** `{remaining_years:.2f}` years"
            )
        else:
            st.error(
                "⚠️ Smoking may have already reduced expected lifespan significantly."
            )

        st.caption(
            "⚠️ This is only an estimate based on averages, not medical advice."
        )

elif smoke == "No":
    st.success("👏 Good choice! Keep it up and maintain a healthy lifestyle.")



```
